```python
import pandas as pd
```

```python
orders = pd.read_csv(
    r"C:\Users\DELL\Downloads\olist\olist_orders_dataset.csv"
)
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
```

```python
orders.dtypes
```

```python
delivered = orders[orders["order_status"] == "delivered"].copy()
```

```python
delivered["is_late"] = (
    delivered["order_delivered_customer_date"]
    > delivered["order_estimated_delivery_date"]
).astype(int)
```

```python
delivered[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "is_late"
    ]
].head(10)
```

```python
delivered["is_late"].value_counts()
```

```python
label_distribution = delivered["is_late"].value_counts(normalize=True) * 100
label_distribution
```

```python
delivered.to_csv("labeled_orders.csv", index=False)
```
